```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
# Read the image
image = cv2.imread(r'C:\Users\moham\Desktop\lab-5\Parrot.png')
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
```

#### TASK-1

```python
# Create a 7x7 box filter
kernel = np.ones((7, 7), dtype=np.float32) / 49

# Convolve the image with the box filter
smoothed = cv2.filter2D(image, -1, kernel)

# Plot original and smoothed images
fig, ax = plt.subplots(1, 2, figsize=(10, 5))

ax[0].imshow(image)
ax[0].set_title("Original Image")
ax[0].axis("off")

ax[1].imshow(smoothed)
ax[1].set_title("Smoothed (7x7 Box Filter)")
ax[1].axis("off")

plt.show()
```

### TASK-2

```python
# Apply Gaussian filters
blurred_5 = cv2.GaussianBlur(image, (5, 5), sigmaX=0)
blurred_21 = cv2.GaussianBlur(image, (21, 21), sigmaX=0)

# Plot original and filtered images
fig, ax = plt.subplots(1, 3, figsize=(15, 5))

ax[0].imshow(image)
ax[0].set_title("Original Image")
ax[0].axis("off")

ax[1].imshow(blurred_5)
ax[1].set_title("Gaussian Filter (5x5)")
ax[1].axis("off")

ax[2].imshow(blurred_21)
ax[2].set_title("Gaussian Filter (21x21)")
ax[2].axis("off")

plt.show()
```

### TASK-3

```python
# Convert image to float32 for processing
image_float = image.astype(np.float32) / 255.0

# Apply a 3x3 Laplacian filter
laplacian = cv2.Laplacian(image_float, cv2.CV_32F, ksize=3)

# Sharpen the image
sharpened = np.clip(image_float - laplacian, 0, 1)

# Prepare the Laplacian result for display
laplacian_display = cv2.normalize(
    laplacian, None, 0, 1, cv2.NORM_MINMAX
)
laplacian_display = np.clip(laplacian_display, 0, 1)

# Plot original, Laplacian, and sharpened images
fig, ax = plt.subplots(1, 3, figsize=(15, 5))

ax[0].imshow(image)
ax[0].set_title("Original Image")
ax[0].axis("off")

ax[1].imshow(laplacian_display)
ax[1].set_title("Laplacian Filter (3x3)")
ax[1].axis("off")

ax[2].imshow(sharpened)
ax[2].set_title("Sharpened Image")
ax[2].axis("off")

plt.show()
```
