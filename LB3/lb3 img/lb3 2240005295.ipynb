Task1: Use any image processing library and apply the following geometric
transformations over the image.
1. Increase the size of the image.
2. Rotate the image by 120 degrees.
3. Perform Sheer operations over the image

```python
from PIL import Image
import matplotlib.pyplot as plt
import numpy as np

# Read the image
img = Image.open(r"C:\Users\M\Desktop\imge\new_image.jpg")

# 1. Increase the size of the image
larger_img = img.resize((img.width * 5, img.height * 5))

# 2. Rotate the image by 120 degrees
rotated_img = img.rotate(120, expand=True)

# 3. Shear operation
shear_factor = 0.3

width, height = img.size

sheared_img = img.transform(
    (width + int(height * shear_factor), height),
    Image.Transform.AFFINE,
    (1, shear_factor, 0, 0, 1, 0)
)

# Display the results
plt.figure(figsize=(12, 8))

plt.subplot(2, 2, 1)
plt.imshow(img)
plt.title("Original Image")
plt.axis("off")

plt.subplot(2, 2, 2)
plt.imshow(larger_img)
plt.title("Increased Size")
plt.axis("off")

plt.subplot(2, 2, 3)
plt.imshow(rotated_img)
plt.title("Rotated 120 Degrees")
plt.axis("off")

plt.subplot(2, 2, 4)
plt.imshow(sheared_img)
plt.title("Sheared Image")
plt.axis("off")

plt.tight_layout()
plt.show()
```

Task2: Use any image processing library and apply the following intensity
transformations over the image.
1. Create a Negative Image
2. Log with (choose an appropriate constant for your image)
3. Power law (choose a value for gamma that increases the contrast)

```python
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image

# Read the image
img = Image.open(r"C:\Users\M\Desktop\imge\lena_gray_256 (1).tif").convert("L")

# Convert image to NumPy array
img_array = np.array(img).astype(float)

# 1. Negative Image
negative = 255 - img_array

# 2. Log Transformation
c = 255 / np.log(1 + np.max(img_array))

log_image = c * np.log(1 + img_array)
log_image = np.uint8(log_image)

# 3. Power Law Transformation
gamma = 0.5

normalized = img_array / 255.0
power_image = 255 * (normalized ** gamma)
power_image = np.uint8(power_image)

# Display the results
plt.figure(figsize=(12, 8))

plt.subplot(2, 2, 1)
plt.imshow(img, cmap="gray")
plt.title("Original Image")
plt.axis("off")

plt.subplot(2, 2, 2)
plt.imshow(negative, cmap="gray")
plt.title("Negative Image")
plt.axis("off")

plt.subplot(2, 2, 3)
plt.imshow(log_image, cmap="gray")
plt.title("Log Transformation")
plt.axis("off")

plt.subplot(2, 2, 4)
plt.imshow(power_image, cmap="gray")
plt.title("Power Law (Gamma = 0.5)")
plt.axis("off")

plt.tight_layout()
plt.show()
```
